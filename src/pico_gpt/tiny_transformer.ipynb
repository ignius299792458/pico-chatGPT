```python
corpus = "the cat sat on the mat. the dog sat on the rug. the cat ate the fish. the dog chased the cat."
len(corpus), corpus
```

```python
# a char is a token
unique_chars = sorted(set(corpus))
unique_chars_len = len(unique_chars)
print(unique_chars)

#  or understand it first 
unique_tokens_in_corpus = unique_chars # since char = token 
vocabulary = unique_tokens_in_corpus
print(vocabulary)
print("total unique chars or vocabulary: ", len(vocabulary))
```

```python
# index mapping to integers of each unique tokens (chars)
stoi = {s: i for i, s in enumerate(unique_chars)}
itos = {i: s for s, i in stoi.items()}
print(stoi, "\n", itos)
```

```python
# training sequence : change each token ( or char) by it index mapped above
datasets_tokens = []

for ch in corpus:
    datasets_tokens.append(stoi[ch])

N = len(datasets_tokens) #number of tokens (characters) in the corpus
print(N, datasets_tokens)
```

```python
T = 8 # context length
```

```python
# context-window sliding through training sequence

X, Y = [], []

for i in range(N-T):
    X.append(datasets_tokens[i:T+i])
    Y.append(datasets_tokens[i+1:T+i+1])

print(X)
print(Y)
```

```python
import torch
```

```python
# Embedding Table
V = len(vocabulary) # vocabulary: total unique characters in corpus: rprt. by V
d = 8 # embedding dimension 

E = torch.randn((V,d))
print(f"{E.shape=}\n{E.dim()=}\n{E=}")
```

```python
Xt, Yt = torch.tensor(X), torch.tensor(Y)
print(f"{len(X)=},{Xt.shape=}")
print(f"{len(Y)=}, {Yt.shape=}")
```

```python
Xte = E[Xt] # Xte = [W, T, d] where W = number of context_windows, T = tokens in a context window and d = embedding dims
print(Xte.shape)
```

```python
Xt[:1].shape, Xte[:1].shape, Xt[:1], Xte[:1]
```

```python
X2 = Xte[0].clone()
X2.shape
```

```python
d_k = d
```

```python
Wq = torch.randn((T, d))
Wk = torch.randn((T, d))
Wv = torch.randn((T, d))
```

```python
# Create Queries Q, Keys K, and Values V
Q = X2 @ Wq
K = X2 @ Wk 
V = X2 @ Wv

print("X :", X2.shape)
print("Q :", Q.shape)
print("K :", K.shape)
print("V :", V.shape)
```

```python
# Attention score matrix S 
scores = Q @ K.T

print("scores :", scores.shape)
print(scores)
```

```python
# Scale the scores
scores = scores / (d_k ** 0.5)
scores
```

```python
# Causal Masking
mask = torch.tril(torch.ones(T,T))
scores = scores.masked_fill(mask == 0, float("-inf"))
scores
```

```python
# Convert scores into attention weights

A = torch.nn.functional.softmax(scores, dim=-1)

print("A :", A.shape)
A
```

```python
# Use attention weights to collect information

O = A @ V

print(f"{A.shape=}\n{A=}")
print(f"{V.shape=}\n{V=}")
print(f"{O.shape=}\n{O=}")
print(f"{A.sum(dim=-1)=}")
```
